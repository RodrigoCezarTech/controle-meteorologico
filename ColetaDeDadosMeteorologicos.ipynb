```python
#https://docs.google.com/spreadsheets/d/1iOGwaZEyzdc7FxYPStaIiA4QJy1IHkB88_EUljYly8c/edit?usp=sharing

import pandas as pd
import requests
import time

endpoint_planilha= "https://docs.google.com/spreadsheets/d/1iOGwaZEyzdc7FxYPStaIiA4QJy1IHkB88_EUljYly8c/export?format=csv"
endpoint_api = "https://api.open-meteo.com/v1/forecast"
```

```python
print('---- Dados Coletados ----')
dados_planilha = pd.read_csv(endpoint_planilha)
display(dados_planilha)
```

```python
for indice, linha in dados_planilha.iterrows():

    nome_local = linha['Nome do local']

    try:
        lat_str = str(linha['Latitude']).replace(',', '.')
        lon_str = str(linha['Longitude']).replace(',', '.')

        if lat_str.count('.') > 1:
            partes = lat_str.split('.')
            lat_str = partes[0] + '.' + ''.join(partes[1:])

        if lon_str.count('.') > 1:
            partes = lon_str.split('.')
            lon_str = partes[0] + '.' + ''.join(partes[1:])

        lat = float(lat_str)
        lon = float(lon_str)

        parametros = {
            'latitude': lat,
            'longitude': lon,
            'current': 'temperature_2m'
        }

        print(f'----- {nome_local} -----')

        resposta = requests.get(endpoint_api, params=parametros).json()

        display(resposta)

    except ValueError:
        print(f'Coordenadas inválidas: {nome_local}')

    print('\n---- Dados Coletados ----\n')
    time.sleep(1)
```
